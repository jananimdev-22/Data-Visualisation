```python
#program 2
#create a 3x3 matrix and access 1st element,last,element,entire 2nd row
import numpy as np
a = np.arange(1, 10).reshape(3, 3)
print(a)
first_element = a[0, 0]
last_element = a[-1, -1]
second_row = a[1, :]
print("\nFirst element:", first_element)
print("Last element:", last_element)
print("Entire 2nd row:", second_row)
```
