```python
#program 1
#create a dataframe containing roll number,name,department,cgpa and display all records
import pandas as pd
data={'roll no':[101,102,103,104,105,106,107,108,109,110],
      'name':['arun','priya','rahul','ajay','pavi','ramesh','janani','darshan','hashu','varsha'],
      'department':['BCA','MCA','BSC','BSC','CS','ML','BA','BA','BCA','MCA'],
      'cgpa':[85,90,78,67,99,77,82,79,90,78]}
df=pd.DataFrame(data)
print(df)
```

```python
#program 1
#create a dataframe and find total marks,avg,max,min

data={'roll no':[101,102,103,104,105,106,107,108,109,110],
      'name':['arun','priya','rahul','ajay','pavi','ramesh','janani','darshan','hashu','varsha'],
      'marks':[85,90,78,67,99,77,82,79,90,78]}
df=pd.DataFrame(data)
print(df)
```

```python
print(df['marks'].sum())
print(df['marks'].mean())
print(df['marks'].max())
print(df['marks'].min())
```
