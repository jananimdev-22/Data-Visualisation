```python
df.isnull()
```

```python
df.isnull().sum()
```

```python
df['sepal_length']=df['sepal_length'].fillna(0)
```

```python
df.isnull().sum()
```

```python
mn=df['sepal_width'].mean()
mn
```

```python
df['sepal_width']=df['sepal_width'].fillna(mn)
```

```python
df.isnull().sum()
```

```python
median=df['petal_length'].median()
median
```

```python
df['petal_length']=df['petal_length'].fillna(median)
```

```python
df.isnull().sum()
```

```python
mode=df['petal_width'].mode()[0]
mode
```

```python
df['petal_width']=df['petal_width'].fillna(mode)
```

```python
df.isnull().sum()
```

```python
df['petal_width']=df['petal_width'].ffill()
df['petal_width']=df['petal_width'].bfill()
df.isnull().sum()
```
