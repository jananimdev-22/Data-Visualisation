```python
import matplotlib.pyplot as plt
```

```python
x=[1,3,5,7,9]
y=[2,4,6,8,10]
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.xlim(1,10)
plt.ylim(2,12)
plt.title("Line Plot Example") 
plt.plot(x,y,color='green',linestyle='dotted',linewidth=3,marker='*',markerfacecolor='orange',markersize=10)
plt.savefig("sample.png")
plt.show()
```

```python
x=[16,22,31,46,52]
y=[98,67,83,100,57]
x1=[13,42,56,31,12,34]
y1=[66,87,21,90,46,80]
plt.title("Student Performance")
plt.xlabel("Roll No")
plt.ylabel("Marks")
plt.xlim(1,50)
plt.ylim(0,100)
plt.plot(x,y,label='line-1',color='pink',linestyle='dashed',linewidth=3,marker='D',markerfacecolor='purple',markersize=6)
plt.plot(x1,y1,label='line-2',color='orange',linestyle='dashdot',linewidth=3,marker='*',markerfacecolor='red',markersize=12)
plt.legend()
plt.show()
```

```python
students=['A','B','C','D','E']
mark=[78,99,56,12,88]
plt.xlabel('Students')
plt.ylabel('Marks')
plt.title('Students Marks in Semester')
plt.bar(students,mark,color='pink',width=0.5,edgecolor='purple',label='StudentsMarks')
plt.legend()
plt.show()
```

```python
hours=[2,3,4,5,6]
mks=[45,66,70,80,90]
plt.scatter(hours,mks,color='brown',marker='*',s=120,label='StudentsMarks')
plt.xlabel('Hours studied')
plt.ylabel('Marks obtained')
plt.legend()
plt.show()
```
