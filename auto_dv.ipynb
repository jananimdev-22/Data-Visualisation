```python
import pandas as pd
```

```python
df = pd.read_csv("c:/users/ibscai04.dgvc-mca/janani/auto_data.csv")
```

```python
df.isnull()
```

```python
df.isnull().sum()
```

```python
df['mpg']=df['mpg'].fillna(0)
df.isnull().sum()
```

```python
mn=df['cylinders'].mean()
mn
```

```python
df['cylinders']=df['cylinders'].fillna(mn)
df.isnull().sum()
```

```python
md=df[' horsepower'].median()
md
```

```python
df[' horsepower']=df[' horsepower'].fillna(md)
df.isnull().sum()
```

```python
mo=df['acceleration'].mode()[0]
mo
```

```python
df['acceleration']=df['acceleration'].fillna(mo)
df.isnull().sum()
```

```python
df['model year']=df['model year'].ffill()
df['origin']=df['origin'].bfill()
df.isnull().sum()
```
